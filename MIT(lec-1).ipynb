#### creating a dense layer from scratch

```python
# building a neural network from scratch using pytorch
import torch
import torch.nn as nn

class MyDenseLayer(nn.Module):
    def __init__(self, input_dim, output_dim):
        super(MyDenseLayer, self).__init__()
        # initialize weights and bias as learnable parameters
        self.w = nn.Parameter(torch.randn(input_dim, output_dim, requires_grad=True))
        self.b = nn.Parameter(torch.randn(1, output_dim, requires_grad=True))

    def forward(self, inputs):
        z = torch.matmul(inputs, self.w) + self.b
        output = torch.sigmoid(z)
        return output
```

```python
import torch

# Create an instance of your custom dense layer
layer = MyDenseLayer(input_dim=3, output_dim=2)

# Create a random input (batch size = 5, input_dim = 3)
x = torch.randn(5, 3)

# Forward pass through your custom layer
output = layer(x)

print("Input shape:", x.shape)
print("Output shape:", output.shape)
print("Output values:\n", output)
```
